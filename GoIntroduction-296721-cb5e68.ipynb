```go
package main

import (
	"fmt"
)

// MyError — кастомная ошибка
type MyError struct {
	Code int
	Msg  string
}

func (e *MyError) Error() string {
	return fmt.Sprintf("Error %d: %s", e.Code, e.Msg)
}

func doSomething() error {
	return &MyError{Code: 404, Msg: "Resource not found"}
}

func main() {
	err := doSomething()
	if e, ok := err.(*MyError); ok {
		fmt.Println("Custom error:", e.Code, e.Msg)
	}
}
```

```go
package main

import "fmt"

func safeFunction() {
	defer func() {
		if r := recover(); r != nil {
			fmt.Println("Recovered from panic:", r)
		}
	}()
	panic("Something bad happened")
}

func main() {
	safeFunction()
	fmt.Println("Program continues...")
}
```

```go
package main

import (
    "fmt"
    "sync"
)

func say(s string, wg *sync.WaitGroup) {
    defer wg.Done() // Нужно вызвать Done как функцию

    for i := 0; i < 5; i++ {
        fmt.Println(s)
    }
}

func main() {
    var wg sync.WaitGroup
    wg.Add(1)

    go say("sdf", &wg) // Передача указателя на WaitGroup

    say("sdg", nil) // Второй вызов без использования WaitGroup

    wg.Wait()
}
```

```go

import "fmt"

func sum(a []int, c chan int) {
    sm := 0
    for _, v := range a {
        sm += v
    }
    c <- sm
}

func main() {
    a := []int{1,2,3,4,5}
    c := make(chan int)

    go sum(a[:len(a)/2], c)
    go sum(a[len(a)/2:], c)

    x,y := <-c, <-c
    fmt.Println(x,y,x+y)
}

main()
```

```go
select — ожидание данных из нескольких каналов
```

```go
package main

import (
	"fmt"
	"time"
)

func main() {
	ch1 := make(chan string)
	ch2 := make(chan string)

	go func() {
		time.Sleep(2 * time.Second)
		ch1 <- "Hello from ch1"
	}()
	go func() {
		time.Sleep(1 * time.Second)
		ch2 <- "Hello from ch2"
	}()

	select {
	case msg1 := <-ch1:
		fmt.Println("Received:", msg1)
	case msg2 := <-ch2:
		fmt.Println("Received:", msg2)
	case <-time.After(3 * time.Second):
		fmt.Println("Timeout!")
	}
}
```

```go
Интерфейсы в Go
Интерфейсы в Go — это мощный инструмент для достижения полиморфизма, то есть
способности объектов разных типов предоставлять одинаковый интерфейс для
взаимодействия. В Go интерфейсы позволяют определить набор методов, которые
должны быть реализованы типами, чтобы они могли считаться &quot;совместимыми&quot; с этим
интерфейсом. В отличие от многих других языков программирования, где классы явно
объявляют реализацию интерфейса, в Go интерфейсы реализуются неявно, что делает
их использование более гибким.
Определение интерфейсов
Интерфейс определяется как набор сигнатур методов, которые должен реализовать тип,
чтобы соответствовать этому интерфейсу.
Пример интерфейса

type Speaker interface {
Speak() string
}
Здесь Speaker — это интерфейс, который требует от типов, его реализующих, наличия
метода Speak, который возвращает строку.
Реализация интерфейсов
Чтобы тип соответствовал интерфейсу, он должен реализовать все методы, которые
определены в этом интерфейсе.
```

```go
Композиция интерфейсов
Интерфейсы могут быть составными, то есть один интерфейс может включать методы
других интерфейсов.
Пример составного интерфейса
```

```go
package main

import (
	"fmt"
)

type Reader interface {
	Read(p []byte) (n int, err error)
}

type Writer interface {
	Write(p []byte) (n int, err error)
}

type ReaderWriter interface {
	Reader
	Writer
}

type MyIO struct{}

func (m *MyIO) Read(p []byte) (int, error) {
	return len(p), nil
}

func (m *MyIO) Write(p []byte) (int, error) {
	return len(p), nil
}

func main() {
	var rw ReaderWriter = &MyIO{} // Используем указатель на MyIO
	data := []byte("HELLO")

	// Чтение
	n, err := rw.Read(data)
	if err != nil {
		fmt.Println("Error reading:", err)
	} else {
		fmt.Println("Read bytes:", n)
	}

	// Запись
	n, err = rw.Write(data)
	if err != nil {
		fmt.Println("Error writing:", err)
	} else {
		fmt.Println("Written bytes:", n)
	}
}
```

```go
Интерфейсы и полиморфизм
Полиморфизм в Go достигается именно через интерфейсы. Вместо того, чтобы
полагаться на классы и наследование, как в традиционных объектно-ориентированных

языках, Go использует интерфейсы для описания поведения, а затем позволяет разным
типам реализовывать это поведение.
Пример полиморфизма
```

```go
package main

import (
    "fmt",
    "math"
)

type Shape interface {
    Area() float64
}

type Circle struct {
    radius float64
}

func (c Circle) Area() float64{
    return math.Pi*c.Radius*c.Radius
}


type Rectangle struct {
    Width, Height float64
}

func (r Rectangle) Area() float64 {
    return c.Widthy * c.Height
}


func PrintArea(s Shape){
    s.Area()
}

func main() {
    c := Circle(Radius: 5)
    r := Rectangle(Width: 4, Height: 6)


    PrintArea(c)
    PrintArea(r)
}
```

Горутины (Goroutines)
Горутина — это легковесный поток выполнения, управляемый Go-руттаймом.
Горутины позволяют выполнять функции параллельно, не создавая при этом полной
операционной системы (как в случае с традиционными потоками). Благодаря этому,
горутины обладают малым потреблением памяти и могут запускаться в больших
количествах (тысячи и более).
Запуск горутины
Чтобы запустить функцию в виде горутины, достаточно перед вызовом функции
использовать ключевое слово go.

```go
package main

import (
	"fmt"
	"sync"
)

func printMessage(message string, wg *sync.WaitGroup) {
	defer wg.Done()
	fmt.Println(message)
}

func main() {
	var wg sync.WaitGroup

	messages := []string{"go1", "go2", "go3"}

	for _, msg := range messages {
		wg.Add(1)
		go printMessage(msg, &wg)
	}
	wg.Wait()
}
```

```go
Синхронизация горутин

Горутины могут работать независимо друг от друга, но часто требуется координация их
работы. Для этого Go предоставляет несколько механизмов, среди которых наиболее
важны каналы (channels).
Каналы (Channels)
Каналы в Go используются для передачи данных между горутинами. Они позволяют
одной горутине отправить данные, а другой — получить их, обеспечивая тем самым
синхронизацию выполнения.
Объявление и использование каналов
```

```go
Буферизованные каналы
По умолчанию каналы неблокирующие, то есть отправка или получение данных будет
блокировать выполнение горутины до тех пор, пока другая горутина не выполнит
соответствующую операцию. Однако Go также поддерживает буферизованные каналы,
которые позволяют отправлять несколько значений без необходимости немедленного
получения.
Пример буферизованного канала
```

```go
ch := make(chan int, 3)
```

```go
package main

import (
    "fmt"
)

func worker(id int, jobs <-chan int, results chan<- int) {
    for j := range jobs {
        fmt.Printf("Worker %d processing job %d\n", id, j)
        results <- j * 2
    }
}

func main() {
    jobs := make(chan int, 5)
    results := make(chan int, 5)

    // Запускаем 3 воркера
    for w := 1; w <= 3; w++ {
        go worker(w, jobs, results)
    }

    // Отправляем 5 заданий
    for j := 1; j <= 5; j++ {
        jobs <- j
    }
    close(jobs)

    // Получаем 5 результатов
    for a := 1; a <= 5; a++ {
        res := <-results
        fmt.Println("Result:", res)
    }
}
```

```go
package main

import (
    "fmt"
    "time"
)

func sendData(ch chan<- string) {
    time.Sleep(2 * time.Second)
    ch <- "hi from senddata"
}

func main() {
    ch := make(chan string)
    go sendData(ch)

    select {
    case msg := <-ch:
        fmt.Println("Received message:", msg)
    case <-time.After(1 * time.Second):
        fmt.Println("Timeout: no message received within 1 second")
    }
}
```

```go
package main

import (
    "fmt"
)

// Определяем пользовательский тип ошибки
type MyError struct {
    Code    int
    Message string
}

// Реализуем метод Error() интерфейса error
func (e *MyError) Error() string {
    return fmt.Sprintf("error %d: %s", e.Code, e.Message)
}

// Функция, которая возвращает ошибку типа *MyError
func doSomething(x int) error {
    return &MyError{
        Code:    400,
        Message: "some error occurred",
    }
}

func main() {
    err := doSomething(10)
    if err != nil {
        fmt.Println(err) // Выведет: error 400: some error occurred
    }
}
```

```go
Использование errors.Is и errors.As
В Go 1.13 были введены функции errors.Is и errors.As для более удобного сравнения
и обработки ошибок.
errors.Is используется для проверки, эквивалентна ли ошибка конкретной известной
ошибке.
go
if errors.Is(err, os.ErrNotExist){
    fmt.Println("sdg")
}


errors.As
go

var myErr *MyError
if errors.As(err,myErr){
    fmt.Println("sdg")
}


```

```go
package main

import (
	"fmt"
)

// Структура ошибки типа MyError
type MyError struct {
	Code    int
	Message string
}

func (e *MyError) Error() string {
	return fmt.Sprintf("Error %d: %s", e.Code, e.Message)
}

// Структура ошибки типа AnotherError
type AnotherError struct {
	Info string
}

func (e *AnotherError) Error() string {
	return fmt.Sprintf("Another error: %s", e.Info)
}

func doSomething() error {
	// Генерация ошибки типа AnotherError
	return &AnotherError{Info: "Some info"}
}

func main() {
	err := doSomething()

	// Проверка нескольких типов ошибок с помощью утверждения типа
	switch e := err.(type) {
	case *MyError:
		// Если ошибка типа MyError
		fmt.Println("MyError:", e.Code, e.Message)
	case *AnotherError:
		// Если ошибка типа AnotherError
		fmt.Println("AnotherError:", e.Info)
	default:
		// Если ошибка другого типа
		fmt.Println("Unknown error:", err)
	}
}
```
